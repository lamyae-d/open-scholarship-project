```python
import pandas as pd

df = pd.read_csv(
    "MetObjects_British.csv",
    sep=";"
)

df.head()
```

```python
df.columns
```

```python
df["Culture"].value_counts().head()
```

```python
british_df = df[df["Culture"].str.contains("British", na=False)]

british_df.head()
```

```python
british_df.shape
```

```python
british_df["Object Name"].value_counts().head(10)
```

```python
british_df["Object Name"].value_counts().head(10).plot(kind="bar")
```

```python
british_df["Department"].value_counts().head(10)
```

```python
british_df["Department"].value_counts().head(10).plot(kind="bar")
```

## Department Analysis

The chart shows that most British objects belong to the Department of European Sculpture and Decorative Arts. The Costume Institute and The American Wing also contain important numbers of British objects. This suggests that British heritage objects are strongly connected to decorative arts, fashion, and historical collections.

## Conclusion

This analysis explored British objects in the Metropolitan Museum dataset. The results show that decorative arts, watches, plates, and fashion-related collections are common within British heritage objects. The analysis also demonstrates how Python and pandas can be used for digital humanities and cultural heritage research.
